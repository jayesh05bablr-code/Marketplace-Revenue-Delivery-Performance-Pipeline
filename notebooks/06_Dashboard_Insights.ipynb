```python

from pyspark.sql import functions as F
import matplotlib.pyplot as plt

CATALOG = "marketplace_catalog"
SCHEMA = "team6_marketplace"

def gold(name):
    return spark.table(f"{CATALOG}.{SCHEMA}.{name}")

print("Dashboard analysis started.")
```

```python

monthly = gold("gold_monthly_category_revenue")

monthly_trend = (
    monthly.groupBy("year_month")
    .agg(F.sum("total_revenue").alias("revenue"))
    .orderBy("year_month")
)

display(monthly_trend)
```

```python

state_revenue = gold("gold_revenue_by_state").orderBy(
    F.desc("total_revenue")
)

display(state_revenue)
```

```python

delivery = gold("gold_delivery_by_state").orderBy(
    F.desc("avg_delay_days")
)

display(delivery)
```

```python

seller = gold("gold_seller_scorecard")

top_sellers = (
    seller.orderBy(F.desc("seller_revenue"))
    .limit(10)
)

display(top_sellers)
```

```python

review_analysis = gold("gold_late_delivery_review_analysis")

display(review_analysis)

```
